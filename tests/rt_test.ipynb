```python
import empylib.waveoptics as wv
import empylib.rad_transfer as rt


cases = {'case1': {'N_back': 10.0, 'nh': 1.0, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},             # air/material with no particles
         'case2': {'N_back': 10.0, 'nh': 1.0, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.5},             # air/material with 50% air particles
         'case3': {'N_back': 10.0, 'nh': 1.0, 'tfilm_test': 0, 'Dp': 0.3, 'fv': 0.5},               # air/material with no thickness
         'case4': {'N_back': 10.0+20j, 'nh': 1.0, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},         # air/metal with no particles
         'case5': {'N_back': 10.0+20j, 'nh': 1.0, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.5},         # air/metal with 50% air particles
         'case6': {'N_back': 10.0+20j, 'nh': 1.0, 'tfilm_test': 0, 'Dp': 0.3, 'fv': 0.5},           # air/metal with no thickness
         'case7': {'N_back': 10.0+50j, 'nh': 1.5+1E-5j, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},    # absorptive/metal with no particles
         'case8': {'N_back': 10.0+50j, 'nh': 1.5+1E-5j, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.5},    # absorptive/metal with 50% absorptive particles
         'case9': {'N_back': 10.0+50j, 'nh': 1.5+1E-5j, 'tfilm_test': 0, 'Dp': 0.3, 'fv': 0.5},      # absorptive/metal with no thickness
         'case10': {'N_back': 1.0, 'nh': 1.5+1E-5j, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},    # absorptive/metal with no particles
         'case11': {'N_back': 1.0, 'nh': 1.5+1E-5j, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.5},    # absorptive/metal with 50% absorptive particles
         'case12': {'N_back': 1.0, 'nh': 1.5+1E-5j, 'tfilm_test': 0, 'Dp': 0.3, 'fv': 0.5},      # absorptive/metal with no thickness
         'case13': {'N_above': 1.0, 'N_back': 1.0, 'nh': 1.0, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},               # uniform non-absorbing medium, finite thickness
         'case14': {'N_above': 1.0, 'N_back': 1.5, 'nh': 1.5, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},               # top dielectric interface only, finite thickness
         'case15': {'N_above': 1.0, 'N_back': 1.5, 'nh': 1.5, 'tfilm_test': 0.0, 'Dp': 0.3, 'fv': 0.5},               # top dielectric interface only, zero thickness
         'case16': {'N_above': 1.0, 'N_back': 1.0, 'nh': 1.5, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},               # air/dielectric/air finite slab
         'case17': {'N_above': 1.0, 'N_back': 1.0, 'nh': 1.5, 'tfilm_test': 0.0, 'Dp': 0.3, 'fv': 0.5},               # air/dielectric/air zero thickness
         'case18': {'N_above': 1.5+1E-5j, 'N_back': 1.5+1E-5j, 'nh': 1.5+1E-5j, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},  # uniform absorptive medium, finite thickness
         'case19': {'N_above': 1.0, 'N_back': 1.5+1E-5j, 'nh': 1.5+1E-5j, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},  # top absorptive interface only, finite thickness
         'case20': {'N_above': 1.0, 'N_back': 1.5+1E-5j, 'nh': 1.5+1E-5j, 'tfilm_test': 0.0, 'Dp': 0.3, 'fv': 0.5},  # top absorptive interface only, zero thickness
         'case21': {'N_above': 1.0, 'N_back': 1.0, 'nh': 1.5+1E-5j, 'tfilm_test': 5.0, 'Dp': 0.3, 'fv': 0.0},         # air/absorptive/air finite slab
         'case22': {'N_above': 1.0, 'N_back': 1.0, 'nh': 1.5+1E-5j, 'tfilm_test': 0.0, 'Dp': 0.3, 'fv': 0.5},         # air/absorptive/air zero thickness
         }
lam_test = 0.5


print(f'|case | interface | incoh_multilayer | adm_sphere |')
print(f'|:---:|:---------:|:----------------:|:----------:|')
for case, values in cases.items():
    N_back = values['N_back']
    nh = values['nh']
    N_above = values.get('N_above', nh)
    tfilm_test = values['tfilm_test']
    Dp = values['Dp']
    fv = values['fv']

    Ri, Ti = wv.incoh_multilayer(wavelength = lam_test,
                                 N_layers = nh,
                                 N_above = N_above,
                                 N_below = N_back,
                                 thickness = tfilm_test*1E3)

    Rj, Tj, *_ = wv.interface(N_above, N_back)

    out_test = rt.adm_sphere(wavelength = lam_test,
                           N_host = nh,
                           N_particle = nh,
                           D = Dp,
                           fv = fv,
                           thickness = tfilm_test,
                           N_above = N_above,
                           N_below = N_back,
                           effective_medium = True,
                           use_phase_fun = True,
                           dependent_scatt = True)
    # print(Ri, Rj, out_test.Rtot[0])
    print(f'|{case} | {Rj[0]:.6f} | {Ri[0]:.6f} | {out_test.Rtot.values[0]:.6f} |')
```
